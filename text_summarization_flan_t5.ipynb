# Text Summarization using Transformers (FLAN-T5)

The objective of this project is to perform text summarization using a pretrained transformer model.
The model used is FLAN-T5, which is an instruction-tuned model capable of understanding and executing tasks based on prompts.

```python
!pip install transformers sentencepiece --quiet
```

```python
from transformers import pipeline
```

Load Model

```python
summarizer = pipeline(
    "text2text-generation",
    model="google/flan-t5-base"
)
```

Input Text

```python
text = """
Artificial Intelligence is rapidly transforming industries across the world.
It is widely used in healthcare for diagnosis, in finance for fraud detection,
and in retail for personalized recommendations. AI improves efficiency, reduces
human error, and enables automation of repetitive tasks. However, it also raises
concerns about job displacement, data privacy, and ethical use.
"""
```

Basic prompt

```python
prompt = "Summarize the following text:\n" + text

result = summarizer(prompt, max_length=40)

print("Basic Output:\n", result[0]['generated_text'])
```

Improved prompt

```python
prompt = "Summarize the following text accurately without adding new information:\n" + text

result = summarizer(prompt, max_length=40, min_length=15, do_sample=False)

print("Improved Output:\n", result[0]['generated_text'])
```

Experimentation

```python
prompts = [
    "Summarize the following text in one sentence:\n",
    "Summarize the following text accurately without adding new information:\n",
    "Provide a detailed summary including key points:\n"
]

for p in prompts:
    result = summarizer(p + text, max_length=50)
    print("\nPrompt:", p)
    print("Output:", result[0]['generated_text'])
```

```python
print("Original Text Length:", len(text.split()))
print("Summary Length:", len(result[0]['generated_text'].split()))
```

# Observations

- Simple prompts often produce overly generic summaries.
- FLAN-T5 tends to generate short and generalized outputs if not guided properly.
- More specific prompts improve the quality and relevance of summaries.
- Detailed prompts help retain important information from the original text.
- Prompt engineering is essential for controlling model behavior.

This project demonstrates how pretrained transformer models can be used for text summarization.
It also highlights the importance of prompt engineering in controlling model behavior and improving output quality.
